# 📘 Supervised Learning – Regression (Improved Version)
### Walmart Weekly Sales Forecasting with Ridge Regression and Log-Transformation
> Improves accuracy for large sales values and avoids underfitting.

```python
# --- Setup and Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# --- Load Merged Walmart Dataset ---
url = "https://raw.githubusercontent.com/saikisri97/Optimisation/master/For_AI_Lecture/data/clean_data_regression.csv"
df = pd.read_csv(url)

# Drop extra index column if present
if 'Unnamed: 0' in df.columns:
    df.drop(columns='Unnamed: 0', inplace=True)
df.head()
```

```python
# --- Select Features and Target ---
features = [
    'Store', 'Dept', 'Fuel_Price', 'IsHoliday', 'CPI', 'Unemployment',
    'Temperature', 'Size', 'month', 'Super_Bowl', 'Labor_Day', 'Thanksgiving', 'Christmas'
]
X = df[features].fillna(0)
X['IsHoliday'] = X['IsHoliday'].astype(int)
y = df['Weekly_Sales']
```

```python
# --- Apply Log Transformation to Target (Improves Prediction of High Sales) ---
y_log = np.log1p(y)  # log(1 + y) handles zeros safely
```

```python
# --- Train/Test Split ---
X_train, X_test, y_train_log, y_test_log = train_test_split(X, y_log, test_size=0.2, random_state=42)
```

```python
# --- Train Ridge Regression Model ---
model = Ridge(alpha=1.0)
model.fit(X_train, y_train_log)
```

```python
# --- Predict and Convert Back to Original Scale ---
y_pred_log = model.predict(X_test)
y_pred = np.expm1(y_pred_log)  # Invert log1p
y_test = np.expm1(y_test_log)
```

```python
# --- Evaluate Model ---
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(np.mean((y_test - y_pred) ** 2))
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.2f}")
```

```python
# --- Visualize Predictions ---
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.3, color='blue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel("Actual Weekly Sales")
plt.ylabel("Predicted Weekly Sales")
plt.title("Walmart Weekly Sales: Actual vs Predicted (Improved)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
