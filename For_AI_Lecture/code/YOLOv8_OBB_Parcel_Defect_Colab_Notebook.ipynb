# 📦 Amazon Parcel Defect Detection with YOLOv8-OBB (Colab Edition)
This notebook is designed for classroom use. It:
- Clones the dataset from GitHub
- Fixes the `data.yaml` to point to local Colab paths
- Trains a YOLOv8 Oriented Bounding Box model on parcel defect data
- Works end-to-end in Google Colab

```
# --- 🔧 Step 1: Install YOLOv8 ---
!pip install ultralytics -q
```

```
# --- 📥 Step 2: Clone the GitHub Dataset ---
!git clone https://github.com/saikisri97/Optimisation.git
```

```
# --- 🛠️ Step 3: Patch data.yaml to use local Colab paths ---
import os

yaml_path = "/content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/data.yaml"
os.makedirs(os.path.dirname(yaml_path), exist_ok=True)

# Overwrite the YAML file (use train for all splits for simplicity)
with open(yaml_path, "w") as f:
    f.write("""train: /content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/train/images
val: /content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/train/images
test: /content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/train/images
nc: 2
names: ['defect parcel', 'no defect parcel']
""")
```

```
# --- 🧠 Step 4: Load YOLOv8-OBB Model ---
from ultralytics import YOLO
model = YOLO("yolov8n-obb.pt")  # Nano model: small and fast
```

```
# --- 🚆 Step 5: Train the Model ---
model.train(
    data="/content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/data.yaml",
    epochs=10,
    imgsz=640,
    batch=8,
    name="parcel_defect_yolo",
    workers=2
)
```

```
# --- 📊 Step 6: Evaluate Model on Test Set ---
metrics = model.val(split='test')
print("Evaluation Metrics:", metrics)
```

```
# --- 🔍 Step 7: Run Inference on Test Images ---
results = model.predict(
    source="/content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/train/images",
    save=True,
    imgsz=640
)

# Visuals will be saved in /runs/predict
```

✅ Done! You can now open `/runs/predict` to see results.
