# Prompt Engineering Demo with FLAN-T5
This notebook demonstrates how to use prompt-based inputs for Q&A using HuggingFace's `flan-t5-base` model.

```python
# Import the required pipeline from HuggingFace Transformers

import tiktoken
from transformers import GPT2Tokenizer, pipeline
```

```python
# Load the FLAN-T5 model for text-to-text generation
qa_pipeline = pipeline("text2text-generation", model="google/flan-t5-base")
```

```python
# Define your question and contextual input
question = "Where is my order with tracking ID 12345?"
context = """
Package 12345 was scanned in Erlangen facility at 9:42 AM.
Expected delivery: Tomorrow by 6 PM.
"""
```

```python
# Format prompts like: "question: <your question> context: <your data>"
input_text = f"question: {question} context: {context}"
```

```python
# ---- Tokenization and Embedding ----
# Use GPT-2 tokenizer with tiktoken
enc = tiktoken.get_encoding("gpt2")
token_ids = enc.encode(input_text)
tokens = [enc.decode_single_token_bytes(tok).decode('utf-8', errors='ignore') for tok in token_ids]

# Print token breakdown
print("🔹 Tokens and IDs:")
for tok, tid in zip(tokens, token_ids):
    print(f"{tok!r} → {tid}")
```

```python
# Generate a response from the model
response = qa_pipeline(input_text)
print('Your order is at -  ',response[0]['generated_text'])
```

> Format prompts like: `question: <your question> context: <your data>`
