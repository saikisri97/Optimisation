# 📦 Amazon Package Defect Detection with YOLOv8 (Oriented Bounding Boxes)
### Using Roboflow's "Parcels-with-Defect-1" dataset

```python
# --- 🔧 Step 1: Install YOLOv8 Library ---
# Install the official Ultralytics package for YOLOv8

!pip install ultralytics -q
```

```python
# --- 📥 Step 2: Import Libraries and Verify Installation ---

from ultralytics import YOLO
import os

# Optional: Check GPU availability
import torch
print("CUDA available:", torch.cuda.is_available())
```

```python
# --- 🧠 Step 3: Load Pretrained YOLOv8-OBB Model ---
# 'obb' = oriented bounding box, useful for tilted or rotated objects like skewed packages

model = YOLO('yolov8n-obb.pt')  # 'n' = nano model (fast, lightweight)
# Option: switch to 'yolov8s-obb.pt' for better accuracy
```

```python
data_path  = '/content/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/data.yaml'
```

```python
# --- 🚆 Step 4: Train the Model on Defect Dataset ---

model.train(
    data=f'{data_path',  # Path to Roboflow YAML file
    epochs=1,              # Increase for better accuracy
    imgsz=640,              # Standard image size
    batch=8,                # Adjust based on GPU RAM
    name='parcel_defect_yolo',
    workers=2,              # Number of data loader workers
)
```

```python
# --- 📊 Step 5: Evaluate Model Performance on Test Set ---

metrics = model.val(split='test')  # Use the test split from Roboflow
print("Test Metrics:", metrics)
```

```python
# --- 🔍 Step 6: Run Inference on Test Images and Visualize Results ---

results = model.predict(
    source='/Users/gollapsi/Documents/08_For_Roland_SG/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1/test/images',  # Folder with test images
    save=True,         # Save annotated results
    imgsz=640,
)

# In Colab, you can browse results in /runs/predict/
```
