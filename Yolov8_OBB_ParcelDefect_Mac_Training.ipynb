# 🧠 Train YOLOv8-OBB on M2 Mac: Parcel Defect Detection
Use this notebook on your **local Mac** to train YOLOv8-OBB model using your dataset, and export a `.pt` checkpoint for use in Colab.

```python
# --- 🔧 Step 1: Install Ultralytics ---
!pip install ultralytics --quiet
```

```python
# --- 📂 Step 2: Define Paths ---
dataset_path = "/Users/gollapsi/Documents/08_For_Roland_SG/Optimisation/For_AI_Lecture/data/Parcels-with-Defect-1"
yaml_path = f"{dataset_path}/data.yaml"

# Create YAML file
with open(yaml_path, "w") as f:
    f.write("""train: {dataset_path}/train/images
val: {dataset_path}/valid/images
test: {dataset_path}/test/images
nc: 2
names: ['defect parcel', 'no defect parcel']
""".replace("{dataset_path}", dataset_path))
```

```python
# --- 🏋️ Step 3: Train YOLOv8n-OBB ---
from ultralytics import YOLO

model = YOLO("yolov8n-obb.pt")  # or yolov8s-obb.pt
model.train(
    data=yaml_path,
    epochs=20,
    imgsz=640,
    batch=8,
    name="parcel_defect_mac",
    workers=2
)
```

```python
# --- 💾 Step 4: Save Trained Model ---
# Trained model will be in runs/obb/parcel_defect_mac/weights/best.pt
# Copy it or upload to GitHub/Drive for use in Colab

import shutil
shutil.copy("runs/obb/parcel_defect_mac/weights/best.pt", "parcel_defect_trained_mac.pt")
```
